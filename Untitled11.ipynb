```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import roc_curve
from sklearn.metrics import auc

from wordcloud import WordCloud

import nltk
import re
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

from hmmlearn import hmm

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input
from tensorflow.keras.layers import Embedding
from tensorflow.keras.layers import Bidirectional
from tensorflow.keras.layers import LSTM
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import Attention
from tensorflow.keras.layers import GlobalAveragePooling1D
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

nltk.download('punkt')
nltk.download('stopwords')
```

```python

fake_df = pd.read_csv('/content/Fake.csv')
true_df = pd.read_csv('/content/True.csv')

fake_df['label'] = 0
true_df['label'] = 1

print(fake_df.head())
print(true_df.head())
```

```python

df = pd.concat([fake_df, true_df], axis=0)

df = df[['text', 'label']]

df = df.sample(frac=1).reset_index(drop=True)

print(df.shape)
```

```python

stop_words = set(stopwords.words('english'))


def clean_text(text):

    text = text.lower()

    text = re.sub(r'https?://\S+', '', text)

    text = re.sub(r'<.*?>', '', text)

    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text)

    text = re.sub(r'\n', '', text)

    text = re.sub(r'\w*\d\w*', '', text)

    tokens = word_tokenize(text)

    filtered = [word for word in tokens if word not in stop_words]

    return ' '.join(filtered)


df['clean_text'] = df['text'].apply(clean_text)

print(df.head())
```

```python

plt.figure(figsize=(8,6))

sns.countplot(x=df['label'])

plt.title('Fake vs Real News Distribution')

plt.xticks([0,1], ['Fake', 'Real'])

plt.show()
```

```python

df['text_length'] = df['clean_text'].apply(len)

plt.figure(figsize=(10,6))

sns.histplot(df['text_length'], bins=50, kde=True)

plt.title('Text Length Distribution')

plt.show()
```

```python

fake_words = ' '.join(df[df['label']==0]['clean_text'])

wordcloud = WordCloud(width=1000,
                      height=500,
                      background_color='white').generate(fake_words)

plt.figure(figsize=(15,7))

plt.imshow(wordcloud)

plt.axis('off')

plt.title('Fake News WordCloud')

plt.show()
```

```python

real_words = ' '.join(df[df['label']==1]['clean_text'])

wordcloud = WordCloud(width=1000,
                      height=500,
                      background_color='black').generate(real_words)

plt.figure(figsize=(15,7))

plt.imshow(wordcloud)

plt.axis('off')

plt.title('Real News WordCloud')

plt.show()
```

```python

X = df['clean_text']

y = df['label']

vectorizer = TfidfVectorizer(max_features=5000)

X_tfidf = vectorizer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y,
    test_size=0.2,
    random_state=42
)
```

```python

X_train_hmm = X_train.toarray()

model_hmm = hmm.GaussianHMM(
    n_components=2,
    covariance_type='diag',
    n_iter=100
)

model_hmm.fit(X_train_hmm)

train_score = model_hmm.score(X_train_hmm)

print('HMM Training Score:', train_score)
```

```python

max_words = 10000
max_len = 300

tokenizer = Tokenizer(num_words=max_words)

tokenizer.fit_on_texts(df['clean_text'])

sequences = tokenizer.texts_to_sequences(df['clean_text'])

X_pad = pad_sequences(sequences, maxlen=max_len)

y = np.array(df['label'])

X_train_dl, X_test_dl, y_train_dl, y_test_dl = train_test_split(
    X_pad,
    y,
    test_size=0.2,
    random_state=42
)
```

```python

max_words = 10000
max_len = 300

tokenizer = Tokenizer(num_words=max_words)

tokenizer.fit_on_texts(df['clean_text'])

sequences = tokenizer.texts_to_sequences(df['clean_text'])

X_pad = pad_sequences(sequences, maxlen=max_len)

y = np.array(df['label'])

X_train_dl, X_test_dl, y_train_dl, y_test_dl = train_test_split(
    X_pad,
    y,
    test_size=0.2,
    random_state=42
)
```

```python

input_layer = Input(shape=(max_len,))

embedding = Embedding(max_words, 128)(input_layer)

bilstm = Bidirectional(LSTM(64, return_sequences=True))(embedding)

attention = Attention()([bilstm, bilstm])

pooling = GlobalAveragePooling1D()(attention)

x = Dense(64, activation='relu')(pooling)

x = Dropout(0.3)(x)

output = Dense(1, activation='sigmoid')(x)

teacher_model = Model(inputs=input_layer, outputs=output)

teacher_model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

teacher_model.summary()
```

```python

history = teacher_model.fit(
    X_train_dl,
    y_train_dl,
    validation_split=0.2,
    epochs=5,
    batch_size=64
)
```

```python

plt.figure(figsize=(10,6))

plt.plot(history.history['accuracy'])

plt.plot(history.history['val_accuracy'])

plt.title('Teacher Model Accuracy')

plt.xlabel('Epoch')

plt.ylabel('Accuracy')

plt.legend(['Train', 'Validation'])

plt.show()
```

```python

plt.figure(figsize=(10,6))

plt.plot(history.history['loss'])

plt.plot(history.history['val_loss'])

plt.title('Teacher Model Loss')

plt.xlabel('Epoch')

plt.ylabel('Loss')

plt.legend(['Train', 'Validation'])

plt.show()
```

```python

student_model = tf.keras.Sequential([

    Embedding(max_words, 64, input_length=max_len),

    LSTM(32),

    Dense(32, activation='relu'),

    Dense(1, activation='sigmoid')

])

student_model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

student_model.summary()
```

```python

student_history = student_model.fit(
    X_train_dl,
    y_train_dl,
    validation_split=0.2,
    epochs=5,
    batch_size=64
)
```

```python

teacher_pred = teacher_model.predict(X_test_dl)

teacher_pred = (teacher_pred > 0.5).astype(int)

teacher_acc = accuracy_score(y_test_dl, teacher_pred)

print('Teacher Accuracy:', teacher_acc)

print(classification_report(y_test_dl, teacher_pred))
```

```python

student_pred = student_model.predict(X_test_dl)

student_pred = (student_pred > 0.5).astype(int)

student_acc = accuracy_score(y_test_dl, student_pred)

print('Student Accuracy:', student_acc)

print(classification_report(y_test_dl, student_pred))
```

```python

cm = confusion_matrix(y_test_dl, teacher_pred)

plt.figure(figsize=(8,6))

sns.heatmap(cm,
            annot=True,
            fmt='d',
            cmap='Blues')

plt.title('Confusion Matrix')

plt.xlabel('Predicted')

plt.ylabel('Actual')

plt.show()
```

```python

fpr, tpr, threshold = roc_curve(y_test_dl, teacher_pred)

roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8,6))

plt.plot(fpr, tpr, label='AUC = %0.2f' % roc_auc)

plt.plot([0,1], [0,1], linestyle='--')

plt.xlabel('False Positive Rate')

plt.ylabel('True Positive Rate')

plt.title('ROC Curve')

plt.legend()

plt.show()
```
